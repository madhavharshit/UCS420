```python
name = "Harshit Pachauri"
age = 20
roll_number = int(input("Enter your roll number: "))
branch = input("Enter your branch: ")
city = input("Enter your home city: ")
cgpa = float(input("Enter your CGPA: "))

digits = [int(d) for d in str(roll_number)]
if len(digits) != 10:
    print("Please enter a 10-digit roll number and rerun this cell.")
else:
    L = [digit * 10 for digit in digits]
    print("L =", L)
```

## 1. Lists

```python
if len(digits) == 10:
    L.append(250)
    print("After append:", L)

    L.insert(2, 350)
    print("After insert:", L)

    L.remove(250)
    print("After remove:", L)

    L.pop(2)
    print("After pop:", L)

    L.sort()
    print("Ascending:", L)

    L.sort(reverse=True)
    print("Descending:", L)

    print("First three:", L[:3])
    print("Last three:", L[-3:])

    average = sum(L) / len(L)
    greater_than_average = [x for x in L if x > average]
    print("Average:", average)
    print("Elements greater than average:", greater_than_average)
```

## 2. Tuples

```python
if len(digits) == 10:
    scores = tuple(L[:8])
    print("scores =", scores)

    highest = max(scores)
    highest_index = scores.index(highest)
    lowest = min(scores)
    lowest_count = scores.count(lowest)

    print("Highest score:", highest)
    print("Index of highest score:", highest_index)
    print("Lowest score:", lowest)
    print("Occurrences of lowest score:", lowest_count)

    reversed_scores = list(reversed(scores))
    print("Reversed tuple as list:", reversed_scores)

    score_input = int(input("Enter a score to search: "))
    if score_input in scores:
        print("First occurrence index:", scores.index(score_input))
    else:
        print("Score not present")

    try:
        scores[0] = 100
    except TypeError as e:
        print("Tuple modification error:", e)
        print("Tuples are immutable, so their elements cannot be changed directly, unlike lists.")

    first_score, second_score, *remaining_scores = scores
    print("First score:", first_score)
    print("Second score:", second_score)
    print("Remaining scores:", remaining_scores)
```

## 3. Random Numbers

```python
import random
from collections import Counter

random.seed(roll_number)
numbers = [random.randint(100, 900) for _ in range(100)]

odd_numbers = [x for x in numbers if x % 2 != 0]
even_numbers = [x for x in numbers if x % 2 == 0]

def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True

prime_numbers = [x for x in numbers if is_prime(x)]
frequency = Counter(numbers)
most_frequent_number, most_frequent_count = frequency.most_common(1)[0]

print("Random numbers:", numbers)
print("Odd count:", len(odd_numbers))
print("Odd numbers:", odd_numbers)
print("Even count:", len(even_numbers))
print("Even numbers:", even_numbers)
print("Prime count:", len(prime_numbers))
print("Prime numbers:", prime_numbers)
print("Most frequent number:", most_frequent_number)
print("Frequency:", most_frequent_count)
```

## 4. Sets

```python
if len(digits) == 10:
    A = {digit * 7 for digit in digits}
    B = {digit * 9 for digit in digits}

    print("A =", A)
    print("B =", B)

    print("Union:", A.union(B))
    print("Intersection:", A.intersection(B))
    print("A - B:", A.difference(B))
    print("B - A:", B.difference(A))
    print("difference() gives directional differences, while symmetric_difference() gives values present in either set but not both.")
    print("Symmetric difference:", A.symmetric_difference(B))
    print("A subset of B:", A.issubset(B))
    print("B superset of A:", B.issuperset(A))

    X = int(input("Enter a value to discard from A: "))
    A.discard(X)
    print("A after discard:", A)
    print("discard() is safer because it does not raise an error when the value is absent.")
```

## 5. Dictionaries

```python
my_dict = {
    "name": name,
    "roll_no": roll_number,
    "branch": branch,
    "age": age,
    "city": city
}

location_value = my_dict.pop("city")
my_dict["location"] = location_value
print("After renaming city to location:", my_dict)

my_dict["cgpa"] = cgpa
print("After adding CGPA:", my_dict)

my_dict["age"] += 1
print("After increasing age:", my_dict)

dict_pop = my_dict.copy()
removed_branch = dict_pop.pop("branch")
print("Using pop():", dict_pop)
print("Value returned by pop():", removed_branch)

dict_del = my_dict.copy()
del dict_del["branch"]
print("Using del:", dict_del)
print("pop() returns the removed value, while del only removes the key-value pair.")

for key, value in my_dict.items():
    print(f"{key} → {value}")

if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key is not present.")

friend_dict = {
    "name": "Rahul Sharma",
    "roll_no": 1234567890,
    "branch": branch,
    "age": 20,
    "city": "Delhi"
}

merged_dict = {**my_dict, **friend_dict}
print("Merged dictionary:", merged_dict)
print("When both dictionaries have the same key, the value from the second dictionary wins.")

string_values = {key: value for key, value in my_dict.items() if isinstance(value, str)}
print("Dictionary containing only string values:", string_values)
```
