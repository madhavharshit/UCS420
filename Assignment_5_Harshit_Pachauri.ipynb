```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1])

# Find unique values and their frequencies
values, counts = np.unique(x, return_counts=True)

# Find the maximum frequency
max_count = np.max(counts)

# Find the values having maximum frequency
most_frequent = values[counts == max_count]

print("Array:", x)
print("Most frequent value:", most_frequent)
print("Frequency:", max_count)

# Find indices of the most frequent value
for value in most_frequent:
    indices = np.where(x == value)[0]
    print("Indices of", value, ":", indices)
```

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)

max_count = np.max(counts)

most_frequent = values[counts == max_count]

print("Array:", y)
print("Most frequent values:", most_frequent)
print("Frequency:", max_count)

for value in most_frequent:
    indices = np.where(y == value)[0]
    print("Indices of", value, ":", indices)
```

##Question 4

```python
Harshit = np.linspace(10, 100, 25)

print("Array:")
print(Harshit)

# Dimensions
print("\nNumber of dimensions:", Harshit.ndim)

# Shape
print("Shape:", Harshit.shape)

# Total number of elements
print("Total number of elements:", Harshit.size)

# Data type
print("Data type:", Harshit.dtype)

# Total bytes consumed
print("Total bytes consumed:", Harshit.nbytes)

# Transpose using reshape()
transpose_reshape = Harshit.reshape(1, 25)

print("\nTranspose using reshape():")
print(transpose_reshape)

# Transpose using T
transpose_T = Harshit.T

print("\nTranspose using T:")
print(transpose_T)
```

```python
ucs420_Harshit = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_Harshit)

# Mean
print("\nMean:", np.mean(ucs420_Harshit))

# Median
print("Median:", np.median(ucs420_Harshit))

# Maximum
print("Maximum:", np.max(ucs420_Harshit))

# Minimum
print("Minimum:", np.min(ucs420_Harshit))

# Unique elements
print("Unique elements:", np.unique(ucs420_Harshit))

# Reshape to 4 rows and 3 columns
reshaped_ucs420_Harshit = ucs420_Harshit.reshape(4, 3)

print("\nReshaped Array (4 x 3):")
print(reshaped_ucs420_Harshit)

# Resize to 2 rows and 3 columns
resized_ucs420_Harshit = np.resize(ucs420_Harshit, (2, 3))

print("\nResized Array (2 x 3):")
print(resized_ucs420_Harshit)
```
