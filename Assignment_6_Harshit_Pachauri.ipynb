## Q1

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected = temperature + 2
fahrenheit = (corrected * 9/5) + 32
above_32 = corrected[corrected > 32]
count_above_32 = np.count_nonzero(corrected > 32)

print("Corrected temperatures (°C):", corrected)
print("Temperatures (°F):", fahrenheit)
print("Readings above 32°C:", above_32)
print("Count above 32°C:", count_above_32)

print("Vectorization applies operations to an entire array without an explicit loop,")
print("while Boolean indexing selects matching values efficiently.")
```

## Q2

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Total steps:", np.sum(steps))
print("Mean steps:", np.mean(steps))
print("Maximum steps:", np.max(steps))
print("Minimum steps:", np.min(steps))
print("Total for each day:", np.sum(steps, axis=0))
print("Total for each user:", np.sum(steps, axis=1))

max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("Maximum position (user index, day index):", max_position)
print("Maximum position (1-based user, day):", (max_position[0] + 1, max_position[1] + 1))
```

## Q3

```python
original = np.array([1, 2, 3, 4, 5, 6])

subset = original[1:5]
subset[0] = 999
print("Original after modifying slice:", original)
print("Subset:", subset)

copied = original[1:5].copy()
copied[0] = 500
print("Original after modifying copy:", original)
print("Copied array:", copied)

matrix = np.arange(1, 13).reshape(3, 4)
print("\nMatrix:\n", matrix)
print("First row:", matrix[0, :])
print("Last row:", matrix[-1, :])
print("Second column:", matrix[:, 1])
print("Rows 1–2 and columns 2–3 (zero-based indices):\n", matrix[1:3, 2:4])

flat = matrix.flatten()
raveled = matrix.ravel()
print("flatten():", flat)
print("ravel():", raveled)

raveled[0] = 100
print("Matrix after changing ravel result:\n", matrix)
flat[1] = 200
print("Matrix after changing flatten result (unchanged by flatten):\n", matrix)

print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

## Q4

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])

print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)

XT = X.T
print("\nTranspose X.T:\n", XT)
print("The transpose swaps rows and columns, so each feature becomes a row.")

XTX = XT @ X
print("\nX.T @ X:\n", XTX)

XTX_inv = np.linalg.inv(XTX)
print("\nInverse of X.T @ X:\n", XTX_inv)

beta = XTX_inv @ XT @ y
print("\nOLS coefficients:", beta)
print("The coefficients give the estimated change in assistance score per unit of")
print("sleep hours, activity level, and stress level, holding the other features fixed.")

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("Predicted assistance score:", predicted_score)
```
